V4 kết hợp 90% điểm v3 với 10% phiếu dự đoán của nonlinear SVM. SVM/NB-SVM, CatBoost, dictionary và TF-IDF đều học từ train. Không có pretrained embedding hoặc mô hình tải sẵn.

Cấu hình đã chọn qua 3-fold grouped CV trên train: v3 0,76512, v4 0,76868. Các điểm này là kết quả thí nghiệm trước, không phải test độc lập hoặc điểm Kaggle. Notebook này train lại từ đầu cấu hình đã chọn; không dùng file dự đoán cache. Trên CPU Colab, quá trình có thể mất khá nhiều phút.
